```python

import pandas as pd

df = pd.read_csv("traffic.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python

import matplotlib.pyplot as plt

print(df.head())

df.plot(figsize=(12, 5))
plt.title("Website Traffic Data")
plt.xlabel("Record number")
plt.ylabel("Page views")
plt.grid(True)
plt.show()
```

```python

import pandas as pd

# Set date as index
df = df.set_index("date").sort_index()

# Create a complete daily date range
df = df.asfreq("D")

# Check missing values after adding missing dates
print("Missing values before filling:")
print(df.isnull().sum())

# Fill missing page views using time-based interpolation
df["views"] = df["views"].interpolate(method="time")

# Fill any remaining missing values using forward/backward filling
df["views"] = df["views"].ffill().bfill()

# Convert page views to non-negative integers
df["views"] = df["views"].clip(lower=0).round().astype(int)

# Restore date as a normal column
df = df.reset_index()

print("\nMissing values after filling:")
print(df.isnull().sum())

print("\nCleaned dataset:")
display(df.head())
```

```python

import pandas as pd
import numpy as np

# Ensure dates are in datetime format
df["date"] = pd.to_datetime(df["date"])

# Sort traffic records from oldest to newest
df = df.sort_values("date").reset_index(drop=True)

# Extract daily page views
traffic = df["views"].to_numpy(dtype=float)

print("First 5 chronological records:")
display(df.head())

print("Last 5 chronological records:")
display(df.tail())

print("Total records:", len(df))
```

```python

# Number of previous days used to predict the next day
sequence_length = 30

X = []
y = []

for i in range(len(traffic) - sequence_length):
    X.append(traffic[i:i + sequence_length])
    y.append(traffic[i + sequence_length])

X = np.array(X)
y = np.array(y)

print("Input sequence shape:", X.shape)
print("Target shape:", y.shape)

print("\nFirst input sequence:")
print(X[0])

print("\nTarget for first sequence:", y[0])
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.plot(range(1, sequence_length + 1), X[0], marker="o")
plt.title("First 30-Day Website Traffic Sequence")
plt.xlabel("Day in Sequence")
plt.ylabel("Page Views")
plt.grid(True)
plt.show()

print("Next-day target:", y[0])
```

```python

split_index = int(len(X) * 0.8)

X_train = X[:split_index]
X_test = X[split_index:]

y_train = y[:split_index]
y_test = y[split_index:]

print("Training sequences:", len(X_train))
print("Testing sequences:", len(X_test))
print("Training targets:", len(y_train))
print("Testing targets:", len(y_test))
```

```python

!pip install -q tensorflow scikit-learn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python

# Fit the scaler only on training data to avoid data leakage
scaler = MinMaxScaler()

train_values = df["views"].iloc[:len(X_train) + 30].to_numpy().reshape(-1, 1)
scaler.fit(train_values)

# Scale input sequences and targets
X_train_scaled = scaler.transform(X_train.reshape(-1, 1)).reshape(X_train.shape)
X_test_scaled = scaler.transform(X_test.reshape(-1, 1)).reshape(X_test.shape)

y_train_scaled = scaler.transform(y_train.reshape(-1, 1))
y_test_scaled = scaler.transform(y_test.reshape(-1, 1))

# Add the single feature dimension
X_train_lstm = X_train_scaled.reshape(X_train_scaled.shape[0], 30, 1)
X_test_lstm = X_test_scaled.reshape(X_test_scaled.shape[0], 30, 1)

print("Training input shape:", X_train_lstm.shape)
print("Testing input shape:", X_test_lstm.shape)
```

```python

model = Sequential([
    LSTM(50, activation="tanh", input_shape=(30, 1)),
    Dropout(0.2),
    Dense(25, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train_lstm,
    y_train_scaled,
    epochs=50,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("LSTM model training completed!")
```

```python

# Predict the held-out test period
pred_scaled = model.predict(X_test_lstm)

predictions = scaler.inverse_transform(pred_scaled).flatten()
actual = scaler.inverse_transform(y_test_scaled).flatten()

predictions = np.maximum(predictions, 0)

results = pd.DataFrame({
    "Actual_Views": actual.astype(int),
    "Predicted_Views": np.round(predictions).astype(int)
})

display(results.head(10))
```

```python

mae = mean_absolute_error(actual, predictions)
rmse = np.sqrt(mean_squared_error(actual, predictions))

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

```python

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Get dates corresponding to the test predictions
test_dates = df["date"].iloc[-len(actual):].reset_index(drop=True)

# Create a comparison DataFrame
comparison = pd.DataFrame({
    "Date": test_dates,
    "Actual_Views": actual,
    "Predicted_Views": predictions
})

# Plot actual vs predicted traffic
plt.figure(figsize=(14, 6))

plt.plot(
    comparison["Date"],
    comparison["Actual_Views"],
    label="Actual Traffic"
)

plt.plot(
    comparison["Date"],
    comparison["Predicted_Views"],
    label="Predicted Traffic"
)

plt.title("Actual vs Predicted Website Traffic (Test Data)")
plt.xlabel("Date")
plt.ylabel("Page Views")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

display(comparison.head(10))
```

```python

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Convert values to NumPy arrays
actual_values = np.asarray(actual).flatten()
predicted_values = np.asarray(predictions).flatten()

# Check that both arrays have the same length
print("Actual values:", len(actual_values))
print("Predicted values:", len(predicted_values))

# Keep only valid pairs
valid = np.isfinite(actual_values) & np.isfinite(predicted_values)
actual_values = actual_values[valid]
predicted_values = predicted_values[valid]

# Calculate accuracy metrics
mae = mean_absolute_error(actual_values, predicted_values)
rmse = np.sqrt(mean_squared_error(actual_values, predicted_values))

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

```python

# Use the most recent 30 days as the starting sequence
last_30 = df["views"].tail(30).to_numpy().reshape(-1, 1)

current_sequence = scaler.transform(last_30).flatten()
future_views = []

# Predict the next 7 days recursively
for day in range(8):
    model_input = current_sequence[-30:].reshape(1, 30, 1)

    next_scaled = model.predict(model_input, verbose=0)[0, 0]

    next_view = scaler.inverse_transform(
        np.array([[next_scaled]])
    )[0, 0]

    next_view = max(0, next_view)
    future_views.append(round(next_view))

    # Append the scaled prediction for the next step
    current_sequence = np.append(current_sequence, next_scaled)

future_dates = pd.date_range(
    start=df["date"].max() + pd.Timedelta(days=1),
    periods=8,
    freq="D"
)

future_df = pd.DataFrame({
    "Date": future_dates,
    "Predicted_Page_Views": future_views
})

print("Predictions for unseen future dates:")
display(future_df)
```

```python

plt.figure(figsize=(12, 5))

plt.plot(
    df["date"].tail(30),
    df["views"].tail(30),
    label="Historical Traffic"
)

plt.plot(
    future_df["Date"],
    future_df["Predicted_Page_Views"],
    marker="o",
    linestyle="--",
    label="Future Predictions"
)

plt.title("Website Traffic Forecast for Unseen Dates")
plt.xlabel("Date")
plt.ylabel("Page Views")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
